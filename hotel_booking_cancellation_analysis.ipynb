## Data Cleaning

```python
#Fill company and agent with 0 ( indicates direct booking)
df['company']=df['company'].fillna(0)
df['agent']=df['agent'].fillna(0)
```

```python
#Fill children with 0
df['children']=df['children'].fillna(0)
```

```python
#Drop the rest of the missing values
df=df.dropna()
```

```python
#Recheck for missing valuues
df.isnull().sum()
```

```python
#Fix data types
#Converting float to integer
df['children']=df['children'].astype(int)
```

```python
#Remove duplicates
df=df.drop_duplicates()
```

```python
#Exporting cleaned data
df.to_csv('cleaned_hotelbooking.csv', index=False)
```

## Feature Engineering

```python
#Calculate total nights
df['total_nights']= df['stays_in_week_nights'] + df['stays_in_weekend_nights']
```

```python
#Calculate total guests
df['total_guests'] = df['adults'] + df['children'] + df['babies']
```

```python
#Create is_family(true if has baby, and false if does not have baby)
df['is_family'] = (df['children']>0) | (df['babies']>0)
```

```python
#Create a seasons column based on the arrival month
month_to_season = {
    'January': 'Winter',
    'February': 'Winter',
    'March': 'Spring',
    'April': 'Spring',
    'May': 'Spring',
    'June': 'Summer',
    'July': 'Summer',
    'August': 'Summer',
    'September': 'Autumn',
    'October': 'Autumn',
    'November': 'Autumn',
    'December': 'Winter'
}

df['seasons'] = df['arrival_date_month'].map(month_to_season)
```
